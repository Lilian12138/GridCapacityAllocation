```python
import arcpy
import os

merged_shp = r"E:\GoldwindProjects2026\SD\processing\gisfiles\scratch\limitedAreasShp\limited_areas_dissolved.shp"
vector_grid = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\MyProject1.gdb\CL_WGS84"
output_gdb = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\limitedArea.gdb"
scratch_gdb = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\scratch.gdb"

arcpy.env.overwriteOutput = True
arcpy.env.workspace = scratch_gdb

# ── 0. 检查输出 GDB 是否存在 ──────────────────────────────────────────
if not arcpy.Exists(output_gdb):
    print("输出 GDB 不存在，正在创建...")
    arcpy.management.CreateFileGDB(os.path.dirname(output_gdb), os.path.basename(output_gdb))
    print("GDB 创建完成：", output_gdb)

# ── 1. 查看网格字段 ───────────────────────────────────────────────────
grid_fields = [f.name for f in arcpy.ListFields(vector_grid)]
print("网格字段列表：", grid_fields)
grid_id_field = "NID10" 

# ── 2. 求交（中间结果写入 scratch_gdb，避免 in_memory 爆内存）────────
intersect_out = rf"{scratch_gdb}\intersect_result"
print("求交运算中...")
arcpy.analysis.Intersect([vector_grid, merged_shp], intersect_out, join_attributes="ALL")
count = int(arcpy.management.GetCount(intersect_out)[0])
print(f"求交完成，共 {count} 个交叉要素")

# 确认求交后实际字段名（OBJECTID 类字段可能带前缀）
intersect_fields = [f.name for f in arcpy.ListFields(intersect_out)]
print("求交后字段列表：", intersect_fields)

# ── 3. 计算 geodesic 面积 ─────────────────────────────────────────────
print("计算 geodesic 面积...")
existing_fields = [f.name for f in arcpy.ListFields(intersect_out)]
if "area_km2" not in existing_fields:
    arcpy.management.AddField(intersect_out, "area_km2", "DOUBLE")
arcpy.management.CalculateGeometryAttributes(
    intersect_out,
    [["area_km2", "AREA_GEODESIC"]],
    area_unit="SQUARE_KILOMETERS"
)

# ── 4. 按网格 ID 汇总面积 ─────────────────────────────────────────────
summary_out = rf"{scratch_gdb}\summary_result"
print("按网格汇总面积中...")
arcpy.analysis.Statistics(
    intersect_out,
    summary_out,
    statistics_fields=[["area_km2", "SUM"]],
    case_field=grid_id_field
)
stat_fields = [f.name for f in arcpy.ListFields(summary_out)]
print("汇总表字段列表：", stat_fields)
sum_field = "SUM_area_km2"  # ← 根据打印结果确认

# ── 5. 复制网格并 Join ────────────────────────────────────────────────
output_fc = rf"{output_gdb}\grid_limited_area"
print("复制原始网格要素...")
arcpy.management.CopyFeatures(vector_grid, output_fc)

print("Join 汇总字段...")
arcpy.management.JoinField(
    in_data=output_fc,
    in_field=grid_id_field,
    join_table=summary_out,
    join_field=grid_id_field,
    fields=[sum_field]
)

# ── 6. 空值填 0 ───────────────────────────────────────────────────────
with arcpy.da.UpdateCursor(output_fc, [sum_field]) as cursor:
    for row in cursor:
        if row[0] is None:
            row[0] = 0
            cursor.updateRow(row)

# ── 7. 统计输出 ───────────────────────────────────────────────────────
print("\n✓ 完成！输出路径：", output_fc)
total = int(arcpy.management.GetCount(output_fc)[0])
print(f"总网格数：{total}")

with arcpy.da.SearchCursor(output_fc, [sum_field]) as cursor:
    values = [row[0] for row in cursor if row[0] and row[0] > 0]

if values:
    print(f"含限制区的网格数：{len(values)}")
    print(f"最大网格限制区面积：{max(values):.2f} km²")
    print(f"含限制区网格平均面积：{sum(values)/len(values):.2f} km²")
    print(f"限制区总面积：{sum(values):.2f} km²")
else:
    print("⚠️ 没有网格与限制区相交，请检查坐标系或数据范围")
```
