# 有效面积计算
## 1. 风光开发潜力计算
### 1.1 风电可开发面积统计
以构建好的网格基准，统计全国所有网格中的用地面积，面积的计算采用的是排除法。风电的排除法包含以下：
a) 排除坡度大于30度的区域

b) 排除植被覆盖、居民区、水体、主干道外扩1000m、机位点外扩500米、光伏板区域

c) 排除海拔高度超过4000米的区域

d) 年平均风速不低于4.5 m/s

```python
import arcpy
import os

arcpy.env.overwriteOutput = True
arcpy.management.ClearWorkspaceCache()

# buffer wind turbine points
input_folder = r"E:\GoldwindProjects2026\SD\processing\gisfiles\风光分布_osm_20260605"
scratch_folder = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\scratch.gdb"
output_folder = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\limitedArea.gdb"

windTurbine_points = os.path.join(input_folder, r"风机点位.shp")
windTurbine_points_buffer = os.path.join(output_folder, r"windTurbine_points_buffer500m")

# 1. Buffer
arcpy.analysis.PairwiseBuffer(
    windTurbine_points,
    windTurbine_points_buffer,
    "500 Meters",
    dissolve_option="ALL",
    method="GEODESIC"
)

# 2. Multipart to singlepart
buffer_singlepart = os.path.join(scratch_folder, r"windTurbine_points_buffer500m_singparts")
arcpy.management.MultipartToSinglepart(
    in_features=windTurbine_points_buffer,
    out_feature_class=buffer_singlepart
)

# 3. Repair geometry
arcpy.management.RepairGeometry(
    in_features=buffer_singlepart,
    delete_null="DELETE_NULL",
    validation_method="OGC"
)

# 4. Add unique ID (creates field via field_type, no separate AddField needed)
arcpy.management.CalculateField(
    buffer_singlepart, "ID", "!OBJECTID!", "PYTHON3", field_type="LONG"
)
```

```python
input_folder = r"E:\GoldwindProjects2026\SD\processing\gisfiles"
scratch_folder = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\scratch.gdb"
output_folder = r"
# buffer road and railways
road_path = r"E:\GoldwindProjects2026\SD\processing\gisfiles\限制性因素\道路\道路.shp"
railways = r"E:\GoldwindProjects2026\SD\processing\gisfiles\限制性因素\铁路\铁路.shp"

# intermediate temp file
roads_railways_merged = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\scratch.gdb\roads_railways_merged"
dissolve_roads_railways = r"E:\GoldwindProjects2026\SD\processing\arcprojects\MyProject1\scratch.gdb\dissolve_roads_railways_buffer100m"

# step 1: merge road and railways into one layer
arcpy.management.Merge(
    inputs=[road_path, railways],
    output=roads_railways_merged
)

# step 2: buffer 1000m and dissolve to single feature
arcpy.analysis.PairwiseBuffer(
    in_features=roads_railways_merged,
    out_feature_class=dissolve_roads_railways,
    buffer_distance_or_field="1000 Meters",
    dissolve_option="ALL",
    method="GEODESIC"
)

# step 3: fix geometry
arcpy.management.RepairGeometry(
    in_features=dissolve_roads_railways,
    delete_null="DELETE_NULL",
    validation_method="OGC"
)
```
